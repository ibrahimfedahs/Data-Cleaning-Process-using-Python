```python
import pandas as pd
```

```python
df = pd.read_csv("heights.csv")
print(df)
```

```python
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
df_clean = df.dropna()
```

```python
df_clean.to_csv("Clean_heights.csv", index=False)
```

```python
print("Clean data saved successfully!")
```

```python
Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)

IQR = Q3 - Q1

df = df[~((df.select_dtypes(include='number') < (Q1 - 1.5 * IQR)) |
          (df.select_dtypes(include='number') > (Q3 + 1.5 * IQR))).any(axis=1)]

print(df)
```

```python
from scipy import stats
import numpy as np

numeric_df = df.select_dtypes(include='number')

z = np.abs(stats.zscore(numeric_df))

df = df[(z < 3).all(axis=1)]

print(df)

# Save final cleaned data
df.to_csv("final_clean_heights.csv", index=False)

print("Final data saved successfully")
```
